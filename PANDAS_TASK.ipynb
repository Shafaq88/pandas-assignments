```python
import pandas as pd
```

```python
#task 1

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nMaximum Marks:")
print(df["Marks"].max())

print("\nMinimum Marks:")
print(df["Marks"].min())

print("\nTotal Students:")
print(len(df))
```

```python
import pandas as pd
#task 2
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)
print(df[df["Marks"] >= 50])

print(df[df["Marks"] < 50])
```

```python
#task 4
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Course": ["AI", "Web", "AI", "Web", "AI"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nCourse Wise Average Marks:")
print(df.groupby("Course")["Marks"].mean())
```

```python
#task 5

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [80, None, 70, None, 90]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nNumber of Missing Values:")
print(df["Marks"].isnull().sum())

average = df["Marks"].mean()

df["Marks"] = df["Marks"].fillna(average)

print("\nCleaned DataFrame:")
print(df)
```
